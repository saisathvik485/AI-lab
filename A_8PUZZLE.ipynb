```python

import heapq

def heuristic(state, goal):
    return sum(1 for i in range(9)
               if state[i] != 0 and state[i] != goal[i])

def astar(start, goal):
    pq = [(heuristic(start, goal), 0, start, [])]
    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state == goal:
            print("Goal reached!")
            print("Total moves:", g)

            for step in path + [state]:
                for i in range(0, 9, 3):
                    print(step[i:i+3])
                print()
            return

        if state in visited:
            continue
        visited.add(state)

        zero = state.index(0)
        row, col = divmod(zero, 3)

        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                pos = r * 3 + c
                new = list(state)
                new[zero], new[pos] = new[pos], new[zero]
                new = tuple(new)

                if new not in visited:
                    heapq.heappush(
                        pq, (g + 1 + heuristic(new, goal),
                             g + 1, new, path + [state])
                    )

    print("No solution found")

start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

astar(start, goal)
```
