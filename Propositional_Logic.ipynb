```python
from itertools import product

# Function to check implication
def implies(a, b):
    return not a or b

# Generate all possible truth assignments
print("P Q R | KB | R | R->P | Q->R")

for P, Q, R in product([False, True], repeat=3):

    # Knowledge Base sentences
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    KB = s1 and s2 and s3

    # Queries
    q1 = R
    q2 = implies(R, P)
    q3 = implies(Q, R)

    # Display truth table
    print(P, Q, R, "|", KB, "|",
          q1, "|", q2, "|", q3)

# Check entailment
models = []

for P, Q, R in product([False, True], repeat=3):

    KB = (implies(Q, P) and
          implies(P, not Q) and
          (Q or R))

    if KB:
        models.append((P, Q, R))

print("\nModels of KB:", models)

print("\nKB entails R:",
      all(R for P, Q, R in models))

print("KB entails R->P:",
      all(implies(R, P) for P, Q, R in models))

print("KB entails Q->R:",
      all(implies(Q, R) for P, Q, R in models))
```
