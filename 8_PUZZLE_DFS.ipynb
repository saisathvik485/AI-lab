```python
def show(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])

def dfs(start, goal, limit=30):
    stack = [(start, 0)]
    parent = {start: None}
    visited = set()

    while stack:
        state, depth = stack.pop()

        if state == goal:
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]
            path.reverse()

            print("Solution found!")
            print("Total moves:", len(path) - 1)

            for i, s in enumerate(path):
                print("\nStep", i)
                show(s)
            return

        if state in visited or depth >= limit:
            continue

        visited.add(state)
        zero = state.index(0)
        row, col = divmod(zero, 3)

        for dr, dc in [(0, 1), (1, 0), (0, -1), (-1, 0)]:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                pos = r * 3 + c
                new = list(state)
                new[zero], new[pos] = new[pos], new[zero]
                new = tuple(new)

                if new not in parent:
                    parent[new] = state
                    stack.append((new, depth + 1))

    print("No solution found within depth limit", limit)


start = (1, 2, 3,
         4, 0, 5,
         6, 7, 8)

goal = (0, 1, 2,
        3, 4, 5,
        6, 7, 8)

dfs(start, goal)
```
