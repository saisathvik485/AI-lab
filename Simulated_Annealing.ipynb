```python

import random
import math

def calculate_cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():
    random.seed(42)

    # Random initial state: one queen per column
    state = [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_cost(state)

    print("Initial State:", [x + 1 for x in state])
    print("Initial Cost:", current_cost)

    temperature = 10.0
    cooling_rate = 0.995
    iteration = 0

    while temperature > 0.01 and current_cost > 0:
        next_state = state.copy()

        # Move one queen to a random row
        col = random.randint(0, 7)
        next_state[col] = random.randint(0, 7)

        next_cost = calculate_cost(next_state)
        delta = next_cost - current_cost

        # Accept better moves or some worse moves
        if delta < 0 or random.random() < math.exp(-delta / temperature):
            state = next_state
            current_cost = next_cost

        temperature *= cooling_rate
        iteration += 1

    if current_cost == 0:
        print("\nGoal State:", [x + 1 for x in state])
        print("Final Cost:", current_cost)
        print("Iterations:", iteration)
        print("\nSolution Found Successfully!")

        print("\nChessboard:")
        for row in range(8):
            for col in range(8):
                if state[col] == row:
                    print("Q", end=" ")
                else:
                    print(".", end=" ")
            print()
    else:
        print("\nSolution not found. Try again.")


simulated_annealing()
```
