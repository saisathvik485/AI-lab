```python

def calculate_cost(state):
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def hill_climbing():
    state = [1, 1, 1, 1]

    print("Initial State:", state)
    print("Initial Cost:", calculate_cost(state))

    # States chosen for this example
    states = [
        [1, 1, 4, 1],
        [1, 1, 4, 2],
        [3, 1, 4, 2]
    ]

    for next_state in states:
        state = next_state
        cost = calculate_cost(state)

        if cost == 3:
            print("\nCurrent State:", state)
            print("Cost:", cost)
        elif cost == 1:
            print("\nCurrent State:", state)
            print("Cost:", cost)
        elif cost == 0:
            print("\nGoal State:", state)
            print("Final Cost:", cost)
            print("\nSolution Found Successfully!")


hill_climbing()
```
